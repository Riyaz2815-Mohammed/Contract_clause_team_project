```python
# View entire memory
print("=" * 50)
print("CONVERSATION MEMORY")
print("=" * 50)
for i, msg in enumerate(memory, 1):
    msg_type = "UESR" if isinstance(msg, HumanMessage) else "BOT"
    print(f"\n[{i}] {msg_type}:\n{msg.content[:100]}...")
```

```python
# Second message - model can see previous messages
user_msg2 = "What did i ate in the breakfast?"
memory.append(HumanMessage(content=user_msg2))

# Send with all previous messages
all_messages = [SystemMessage(content="You are helpful.")] + memory[:-1]  # Exclude the last message we just added

# Last five conversation only given to model
response2 = model.invoke(memory[-5:] + [HumanMessage(content=user_msg2)])
memory.append(response2)

print("User:", user_msg2)
print("Bot:", response2.content)
print()
```

```python
# Create memory list to store conversation
memory = []

# First message
user_msg1 = "Who are the top Tamil actresses?"
memory.append(HumanMessage(content=user_msg1))
response1 = model.invoke([SystemMessage(content="You are helpful."), HumanMessage(content=user_msg1)])
memory.append(response1)
print("User:", user_msg1)
print("Bot:", response1.content)
print()
```

# Simple AI Chatbot with Memory

```python
# Keep only last 5 messages
last_5 = memory[-5:] if len(memory) > 5 else memory

print("=" * 50)
print("LAST 5 CONVERSATIONS")
print("=" * 50)
for i, msg in enumerate(last_5, 1):
    msg_type = "USER" if isinstance(msg, HumanMessage) else "BOT"
    print(f"\n[{i}] {msg_type}:\n{msg.content}")
```
